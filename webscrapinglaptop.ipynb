```python
import requests
import re
from bs4 import BeautifulSoup


brand=[]
Brand_title=[]
processor=[]
ram=[]
ssd=[]
display=[]
price=[]
original_price=[]
discount=[]
ratings=[]
headers={"User-Agent":"Mozilla/5.0"}
for page in range(1,11):
    url=f"https://www.flipkart.com/search?q=laptops&as=on&as-show=on&otracker=AS_Query_TrendingAutoSuggest_1_0_na_na_na&otracker1=AS_Query_TrendingAutoSuggest_1_0_na_na_na&as-pos=1&as-type=HISTORY&suggestionId=laptops&requestId=8809d76a-c51d-49f1-b6a9-23641fb8f697&page={page}"

    response = requests.get(url,headers=headers)
    soup = BeautifulSoup(response.content,"html.parser")
    for tag in soup.find_all("div",class_="RG5Slk"):
        text=tag.text
        a=re.findall(r"(.+)\s\-",text)
        brand.extend(a)
        
    for tag in soup.find_all("div",class_="RG5Slk"):
        text=tag.text
        a= re.findall(r"^[A-Za-z]+", text)
        Brand_title.extend(a)
    for tag in soup.find_all("li",class_="DTBslk"):
        text=tag.text
        a=re.findall(r"(.+)",text)
        processor.extend(a)
        
    for tag in soup.find_all("li",class_="DTBslk"):
        text=tag.text
        a=re.findall(r"(\d+).+RAM",text)
        ram.extend(a)
    
    for tag in soup.find_all("li",class_="DTBslk"):
        text=tag.text
        a=re.findall(r"(\d+)\s.+",text)
        ssd.extend(a)
    for tag in soup.find_all("li",class_="DTBslk"):
        text=tag.text
        a=re.findall(r"(\d+)\s.+",text)
        display.extend(a)
    for tag in soup.find_all("div",class_="hZ3P6w DeU9vF"):
        text=tag.text
        a=re.findall(r"₹(.+)",text)
        price.extend(a)
    for tag in soup.find_all("div",class_="kRYCnD gxR4EY"):
        text=tag.text
        a=re.findall(r"₹(.+)", text)
        original_price.extend(a)
    for tag in soup.find_all('span'):
        text=tag.text
        a=re.findall(r"(\d+)\%\s",text)
        discount.extend(a)
    for tag in soup.find_all("div",class_="MKiFS6"):
        text=tag.text
        a=re.findall(r"(.+)",text)
        ratings.extend(a)
    print(f"page {page} is scraped")
```

```python
print(brand, Brand_title,processor,ram,ssd,display,price,original_price,discount,ratings)
```
